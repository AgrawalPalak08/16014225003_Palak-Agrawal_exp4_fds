```python

import pandas as pd

filename = list(uploaded.keys())[0]

df = pd.read_csv(
    filename,
    usecols=["t2m", "d2m", "u10", "v10", "msl"]
)

print("Dataset imported successfully!")
print(df.head())
print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())
```

```python
import math
import matplotlib.pyplot as plt
```

```python
# Select two columns
data = df[["t2m", "d2m"]].dropna()

x = data["t2m"].tolist()
y = data["d2m"].tolist()

n = len(x)

# Initialize variables
sum_x = 0
sum_y = 0
sum_xy = 0
sum_x2 = 0
sum_y2 = 0

# Calculate sums manually
for i in range(n):
    sum_x = sum_x + x[i]
    sum_y = sum_y + y[i]
    sum_xy = sum_xy + x[i] * y[i]
    sum_x2 = sum_x2 + x[i] * x[i]
    sum_y2 = sum_y2 + y[i] * y[i]

# Calculate numerator and denominator
numerator = n * sum_xy - sum_x * sum_y

a = n * sum_x2 - sum_x * sum_x
b = n * sum_y2 - sum_y * sum_y

denominator_squared = a * b

if denominator_squared > 0:
    denominator = math.sqrt(denominator_squared)
    r = numerator / denominator

    print("Variable X: t2m")
    print("Variable Y: d2m")
    print("Number of records:", n)
    print("Correlation coefficient:", r)

    if r > 0:
        print("Positive correlation")
    elif r < 0:
        print("Negative correlation")
    else:
        print("No linear correlation")
else:
    print("Correlation cannot be calculated.")
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(x, y)

plt.xlabel("Temperature at 2 metres (t2m)")
plt.ylabel("Dew-point temperature (d2m)")
plt.title("Correlation between Temperature and Dew-point Temperature")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import math

x = [1, 2, 3, 4, 5]

data = [
    ([2, 4, 6, 8, 10], "Positive Correlation"),
    ([10, 8, 6, 4, 2], "Negative Correlation"),
    ([3, 1, 5, 1, 3], "No Linear Correlation")
]

for y, title in data:
    n = len(x)
    sx = sy = sxy = sx2 = sy2 = 0

    for i in range(n):
        sx += x[i]
        sy += y[i]
        sxy += x[i] * y[i]
        sx2 += x[i] ** 2
        sy2 += y[i] ** 2

    num = n * sxy - sx * sy
    den = math.sqrt((n * sx2 - sx**2) * (n * sy2 - sy**2))
    r = num / den

    print(title, ":", round(r, 2))

    plt.figure()
    plt.scatter(x, y)
    plt.title(title)
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.grid()
    plt.show()
```
